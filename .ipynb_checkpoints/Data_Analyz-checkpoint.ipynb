```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Đọc dữ liệu từ file CSV
df = pd.read_csv('Advertising.csv')

# 2. Tính toán ma trận tương quan (mặc định là phương pháp Pearson)
correlation_matrix = df.corr()

# In kết quả ra màn hình (dạng bảng số)
print("Ma trận tương quan giữa các biến:")
print(correlation_matrix)

# 3. Trực quan hóa ma trận tương quan bằng Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".3f", linewidths=0.5)

# Cài đặt tiêu đề
plt.title('Ma trận tương quan giữa các loại chi phí quảng cáo và Doanh thu', fontsize=14)

# Hiển thị biểu đồ
plt.show()
```

### Biểu đồ check tương quan với Sales

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Đọc dữ liệu từ file CSV
df = pd.read_csv('Advertising.csv')

# 2. Thiết lập kích thước tổng thể cho khung đồ thị
plt.figure(figsize=(18, 5))

# 3. Vẽ Scatter plot cho TV vs Sales
plt.subplot(1, 3, 1) # (1 hàng, 3 cột, vị trí số 1)
sns.scatterplot(data=df, x='TV', y='Sales', color='blue', alpha=0.7)
plt.title('Mối quan hệ giữa chi phí quảng cáo TV và Doanh thu')
plt.xlabel('Chi phí quảng cáo trên TV')
plt.ylabel('Doanh thu (Sales)')

# 4. Vẽ Scatter plot cho Radio vs Sales
plt.subplot(1, 3, 2) # (1 hàng, 3 cột, vị trí số 2)
sns.scatterplot(data=df, x='Radio', y='Sales', color='green', alpha=0.7)
plt.title('Mối quan hệ giữa chi phí quảng cáo Radio và Doanh thu')
plt.xlabel('Chi phí quảng cáo trên Radio')
plt.ylabel('Doanh thu (Sales)')

# 5. Vẽ Scatter plot cho Newspaper vs Sales
plt.subplot(1, 3, 3) # (1 hàng, 3 cột, vị trí số 3)
sns.scatterplot(data=df, x='Newspaper', y='Sales', color='orange', alpha=0.7)
plt.title('Mối quan hệ giữa chi phí quảng cáo Báo chí và Doanh thu')
plt.xlabel('Chi phí quảng cáo trên Báo chí (Newspaper)')
plt.ylabel('Doanh thu (Sales)')

# 6. Điều chỉnh khoảng cách giữa các đồ thị và hiển thị
plt.tight_layout()
plt.show()
```
