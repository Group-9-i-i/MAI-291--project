### Only TV

```python
import numpy as np 
import pandas as pd 
df = pd.read_csv('Advertising.csv')
X = np.c_[df['TV']]
X = np.c_[np.ones((len(X),1)),df['TV']]
y = df['Sales'].values
Y = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)
print(Y)
```

### Biểu đồ linear regression giữa TV và Sales

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd 
df = pd.read_csv('Advertising.csv')
X_tv = np.c_[df['TV']]
Y_sales = np.c_[df['Sales']]
Sales = Y[0]+Y[1]*X_tv
TV_mse=np.mean((Y_sales-Sales)**2)
SSr= np.mean((Y_sales-Sales)**2)
SSt = np.mean((Y_sales-np.mean(Y_sales))**2)
TV_R2 = 1- (SSr/SSt)
n,k = map(int, X_tv.shape)
TV_R2_Adjusted = 1-((1-TV_R2)*(n-1))/(n-k-1)
print(f'MSE: {TV_mse}')
print(f'R2: {TV_R2}')
print(f'Adjusted R2: {TV_R2_Adjusted}')
plt.figure(figsize=(8, 8)) 
plt.scatter(Y_sales, Sales, alpha=0.6, color='blue', edgecolors='k')

max_val = max(max(Y_sales), max(Sales))
min_val = min(min(Y_sales), min(Sales))
plt.plot([min_val, max_val], [min_val, max_val], color='red', linestyle='--', linewidth=2, label='Đường lý tưởng (Thực tế = Dự đoán)')

plt.xlabel('Doanh Số Thực Tế (Actual Sales)')
plt.ylabel('Doanh Số Dự Đoán (Predicted Sales)')
plt.title('Biểu đồ Phân tán: Doanh số Thực tế vs Dự đoán')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

### TV + Radio

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
df = pd.read_csv('Advertising.csv')
X_tv_radio = np.c_[df['TV'],df['Radio']]
X_tv_radio = np.c_[np.ones((len(X_tv_radio),1)),df['TV'],df['Radio']]
Y_sales = df['Sales'].values
Y_ans= np.linalg.inv(X_tv_radio.T.dot(X_tv_radio)).dot(X_tv_radio.T).dot(Y_sales)
print(Y_ans)
```

### Biểu đồ linear regression giữa TV, Radio và Sales

```python
sales_thucte = df['Sales'].values
X_tvs = df['TV']
X_radio = df['Radio']
sales_dudoan = Y_ans[0]+Y_ans[1]*X_tvs + Y_ans[2]*X_radio
TV_Radio_mse=np.mean((sales_thucte-sales_dudoan)**2)
SSr= np.mean((sales_thucte-sales_dudoan)**2)
SSt = np.mean((sales_thucte-np.mean(sales_thucte))**2)
TV_Radio_R2 = 1- (SSr/SSt)
n,k = map(int, X_tv_radio.shape)
print(k)
TV_Radio_R2_Adjusted = 1-(((1-TV_Radio_R2)*(n-1))/(n-k-1-1))
print(f'MSE: {TV_Radio_mse}')
print(f'R2: {TV_Radio_R2}')
print(f'Adjusted R2: {TV_Radio_R2_Adjusted}')
plt.figure(figsize=(8, 8)) 
plt.scatter(sales_thucte, sales_dudoan, alpha=0.6, color='blue', edgecolors='k')

max_val = max(max(sales_thucte), max(sales_dudoan))
min_val = min(min(sales_thucte), min(sales_dudoan))
plt.plot([min_val, max_val], [min_val, max_val], color='red', linestyle='--', linewidth=2, label='Đường lý tưởng (Thực tế = Dự đoán)')

plt.xlabel('Doanh Số Thực Tế (Actual Sales)')
plt.ylabel('Doanh Số Dự Đoán (Predicted Sales)')
plt.title('Biểu đồ Phân tán: Doanh số Thực tế vs Dự đoán')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

### Biểu đồ linear regression giữa TV, Radio, Newspaper và Sales

```python
sales_thuctes = df['Sales'].values
X_tvss = df['TV']
X_radios = df['Radio']
X_newspaper = df['Newspaper']
sales_dudoans = Y_answ[0]+Y_answ[1]*X_tvss + Y_answ[2]*X_radios+ Y_answ[3]*X_newspaper
TV_Radio_Newspaper_mse=np.mean((sales_thuctes-sales_dudoans)**2)
SSr= np.mean((sales_thuctes-sales_dudoans)**2)
SSt = np.mean((sales_thuctes-np.mean(sales_thuctes))**2)
TV_Radi0_Newspaper_R2 = 1- (SSr/SSt)
n,k = map(int, X_tv_radio_newspaper.shape)
TV_Radio_Newspaper_R2_Adjusted = 1-(((1-TV_Radi0_Newspaper_R2)*(n-1))/(n-1-k-1))
print(f'MSE: {TV_Radio_Newspaper_mse}')
print(f'R2: {TV_Radi0_Newspaper_R2}')
print(f'Adjusted R2: {TV_Radio_Newspaper_R2_Adjusted}')
plt.figure(figsize=(8, 8)) 
plt.scatter(sales_thuctes, sales_dudoans, alpha=0.6, color='blue', edgecolors='k')

max_val = max(max(sales_thuctes), max(sales_dudoans))
min_val = min(min(sales_thuctes), min(sales_dudoans))
plt.plot([min_val, max_val], [min_val, max_val], color='red', linestyle='--', linewidth=2, label='Đường lý tưởng (Thực tế = Dự đoán)')

plt.xlabel('Doanh Số Thực Tế (Actual Sales)')
plt.ylabel('Doanh Số Dự Đoán (Predicted Sales)')
plt.title('Biểu đồ Phân tán: Doanh số Thực tế vs Dự đoán')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

## bài báo cáo đánh giá mcứ độ mạnh yếu của các biến TV, Radio, Newspaper đối với Sales
- Từ các phân tích trên chúng ta có thể thấy rằng TV có mối quan hệ mạnh nhất với Sales, tiếp theo là Radio và cuối cùng là Newspaper. Điều này được thể hiện qua hệ số tương quan và độ dốc của các đường hồi quy trong các biểu đồ linear regression. TV có hệ số tương quan cao nhất và độ dốc lớn nhất, cho thấy rằng khi chi tiêu cho TV tăng lên, Sales cũng tăng lên một cách đáng kể. Radio cũng có mối quan hệ tích cực với Sales, nhưng không mạnh bằng TV. Trong khi đó, Newspaper có mối quan hệ yếu nhất với Sales, cho thấy rằng chi tiêu cho Newspaper không ảnh hưởng nhiều đến doanh số bán hàng.
- Tiếp theo là chỉ số p-value của các biến TV, Radio và Newspaper. Chỉ số p-value cho biết mức độ tin cậy của mối quan hệ giữa các biến độc lập (TV, Radio, Newspaper) và biến phụ thuộc (Sales). Nếu p-value nhỏ hơn 0.05, chúng ta có thể kết luận rằng mối quan hệ giữa biến độc lập và biến phụ thuộc là có ý nghĩa thống kê. Trong trường hợp này, TV có p-value rất nhỏ, cho thấy mối quan hệ giữa TV và Sales là rất mạnh và đáng tin cậy. Radio cũng có p-value nhỏ, nhưng không mạnh bằng TV. Trong khi đó, Newspaper có p-value lớn hơn 0.05, cho thấy rằng mối quan hệ giữa Newspaper và Sales không có ý nghĩa thống kê.
