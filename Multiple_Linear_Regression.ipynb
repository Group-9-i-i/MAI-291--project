### Only TV

```python
import numpy as np 
import pandas as pd 
df = pd.read_csv('Advertising.csv')
X = np.c_[df['TV']]
X = np.c_[np.ones((len(X),1)),df['TV']]
y = df['Sales'].values
Y = np.linalg.inv(X.T.dot(X)).dot(X.T).dot(y)
print(Y)
```

###Nếu TV tăng 1 đơn vị chi tiêu sales tăng trung bình 0.05546477 , tại sao lại dòng cái linalg mà không dùng gradient dencent (Cần giải thích)

### Biểu đồ linear regression giữa TV và Sales

```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd 
df = pd.read_csv('Advertising.csv')
X_tv = np.c_[df['TV']]
Y_sales = np.c_[df['Sales']]

Sales = Y[0]+Y[1]*X_tv
TV_mse=np.mean((Y_sales-Sales)**2)
SSr= np.mean((Y_sales-Sales)**2)
SSt = np.mean((Y_sales-np.mean(Y_sales))**2)
TV_R2 = 1- (SSr/SSt)
n,k = map(int, X_tv.shape)
TV_R2_Adjusted = 1-((1-TV_R2)*(n-1))/(n-k-1)
print(f'MSE: {TV_mse}')
print(f'R2: {TV_R2}')
print(f'Adjusted R2: {TV_R2_Adjusted}')
plt.figure(figsize=(8, 8)) 
plt.scatter(Y_sales, Sales, alpha=0.6, color='blue', edgecolors='k')

max_val = max(max(Y_sales), max(Sales))
min_val = min(min(Y_sales), min(Sales))
plt.plot([min_val, max_val], [min_val, max_val], color='red', linestyle='--', linewidth=2, label='Đường lý tưởng (Thực tế = Dự đoán)')

plt.xlabel('Doanh Số Thực Tế (Actual Sales)')
plt.ylabel('Doanh Số Dự Đoán (Predicted Sales)')
plt.title('Biểu đồ Phân tán: Doanh số Thực tế vs Dự đoán')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

### TV + Radio

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
df = pd.read_csv('Advertising.csv')
X_tv_radio = np.c_[df['TV'],df['Radio']]
X_tv_radio = np.c_[np.ones((len(X_tv_radio),1)),df['TV'],df['Radio']]
Y_sales = df['Sales'].values
Y_ans= np.linalg.inv(X_tv_radio.T.dot(X_tv_radio)).dot(X_tv_radio.T).dot(Y_sales)
print(Y_ans)
```

### Biểu đồ linear regression giữa TV, Radio và Sales

```python
sales_thucte = df['Sales'].values
X_tvs = df['TV']
X_radio = df['Radio']
sales_dudoan = Y_ans[0]+Y_ans[1]*X_tvs + Y_ans[2]*X_radio
TV_Radio_mse=np.mean((sales_thucte-sales_dudoan)**2)
SSr= np.mean((sales_thucte-sales_dudoan)**2)
SSt = np.mean((sales_thucte-np.mean(sales_thucte))**2)
TV_Radio_R2 = 1- (SSr/SSt)
n,k = map(int, X_tv_radio.shape)
print(k)
TV_Radio_R2_Adjusted = 1-(((1-TV_Radio_R2)*(n-1))/(n-k-1-1))
print(f'MSE: {TV_Radio_mse}')
print(f'R2: {TV_Radio_R2}')
print(f'Adjusted R2: {TV_Radio_R2_Adjusted}')
plt.figure(figsize=(8, 8)) 
plt.scatter(sales_thucte, sales_dudoan, alpha=0.6, color='blue', edgecolors='k')

max_val = max(max(sales_thucte), max(sales_dudoan))
min_val = min(min(sales_thucte), min(sales_dudoan))
plt.plot([min_val, max_val], [min_val, max_val], color='red', linestyle='--', linewidth=2, label='Đường lý tưởng (Thực tế = Dự đoán)')

plt.xlabel('Doanh Số Thực Tế (Actual Sales)')
plt.ylabel('Doanh Số Dự Đoán (Predicted Sales)')
plt.title('Biểu đồ Phân tán: Doanh số Thực tế vs Dự đoán')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

## Giải thích các chỉ số đánh giá mô hình

Trong bài toán hồi quy, chúng ta thường sử dụng một số chỉ số để đánh giá mức độ chính xác của mô hình dự đoán. Trong bài này, ba chỉ số được sử dụng là **MSE**, **R²** và **Adjusted R²**.

---

### 1. MSE (Mean Squared Error)

**MSE** là sai số trung bình bình phương giữa giá trị dự đoán và giá trị thực tế.

Nói đơn giản:
- MSE đo **mức độ sai lệch của dự đoán so với dữ liệu thật**.
- Giá trị MSE **càng nhỏ thì mô hình càng chính xác**.

Ví dụ:  
Nếu mô hình dự đoán doanh số rất gần với doanh số thực tế thì MSE sẽ nhỏ.  
Nếu dự đoán sai nhiều thì MSE sẽ lớn.

---

### 2. R² (R-squared)

**R²** cho biết mô hình giải thích được **bao nhiêu phần trăm sự biến thiên của dữ liệu**.

Nói đơn giản:
- R² càng **gần 1** thì mô hình càng tốt.
- R² càng **gần 0** thì mô hình dự đoán kém.

Ví dụ:
- R² = 0.90 nghĩa là mô hình giải thích được **90% sự thay đổi của Sales** dựa trên các biến quảng cáo.

---

### 3. Adjusted R²

**Adjusted R²** là phiên bản cải tiến của R² khi mô hình có nhiều biến.

Vấn đề của R²:
- Khi thêm biến mới vào mô hình, **R² thường sẽ tăng**, kể cả khi biến đó không thực sự quan trọng.

Adjusted R² sẽ:
- **phạt những biến không cần thiết**
- giúp đánh giá mô hình **công bằng hơn khi so sánh các mô hình có số biến khác nhau**

Do đó:
- Khi so sánh nhiều mô hình hồi quy, **Adjusted R² thường đáng tin cậy hơn R²**.

---
