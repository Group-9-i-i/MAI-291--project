# 1. TỔNG QUAN DỮ LIỆU (INTRODUCTION & DATASET)
**Mục tiêu:** Tải dữ liệu, kiểm tra các đặc trưng cơ bản và xem xét thống kê mô tả để có cái nhìn đầu tiên về bài toán phân bổ ngân sách quảng cáo.


```python
# Import thư viện cần thiết
import pandas as pd

# Tải dữ liệu
df = pd.read_csv('Advertising.csv')

# 1. Kích thước và thông tin dữ liệu
print("--- TỔNG QUAN DỮ LIỆU ---")
print(f"Số dòng (Observations): {df.shape[0]}")
print(f"Số cột (Features + Target): {df.shape[1]}")
print("\n--- KIỂU DỮ LIỆU VÀ GIÁ TRỊ THIẾU ---")
df.info() # Lệnh này hiển thị cả kiểu dữ liệu và số lượng non-null
```

*Nhận xét: Dữ liệu sạch (0 giá trị Null) và hoàn toàn ở định dạng số liên tục (float64). Sẵn sàng cho việc phân tích.*


```python
# 2. Xem thống kê mô tả
print("\n--- THỐNG KÊ MÔ TẢ (Descriptive Statistics) ---")
display(df.describe().T) # Dùng .T (Transpose) để bảng xoay ngang, dễ nhìn và chuyên nghiệp hơn
```

```python
# 3. Xem trước 5 dòng đầu tiên
print("\n--- XEM TRƯỚC DỮ LIỆU (Head) ---")
display(df.head())
```
